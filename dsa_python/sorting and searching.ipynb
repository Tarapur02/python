```python
# Selection Sort
def selection_sort(arr):
    n = len(arr)

    for i in range(n-1):
        # Assume the current element is the minimum
        min_idx = i

        # Find the index of the minimum element in the unsorted part
        for j in range(i+1 , n):
            if arr[min_idx] > arr[j]: # Corrected: Compare values to find the actual minimum
                min_idx = j         # Corrected: Store the index, not the value

        # Swap the found minimum element with the first element of the unsorted part
        temp = arr[i]
        arr[i] = arr[min_idx]
        arr[min_idx] = temp

    return arr

arr = [64, 25, 12, 22, 11]
print("Original:", arr)
print("Sorted:", selection_sort(arr.copy()))
```
